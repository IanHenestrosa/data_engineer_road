### **1. try / except / else / finally — qué hace cada bloque y cuándo usarlo**

````python
try:
    data = pd.read_csv(wrong_bios_dir)
except FileNotFoundError as e:
    print(f"Error: {e}")
else:
    print("Data loaded successfully.") #Este else no se ejecuto porque entro al except
finally:
    print("Execution completed.") # Finally se ejecuta siempre, sin importar si hubo un error o no.
````

### **2. Excepcion personalizada. Hereda de la Clase Exception**
Una exepcion personalizada sigue este patron

```python
class MiError(Exception):
    pass
# Luego
raise MiError("Descripción del problema")

# Por ultimo
try:
    ...
except MiError as e:
    ...
```

Ejemplo basico

````python
class SaldoInsuficienteError(Exception):
    # Una excepcion puede tener distintos datos.
    def __init__(self, saldo_actual, retiro):
        self.saldo_actual = saldo_actual
        self.retiro = retiro

        super().__init__(
            f"Saldo insuficiente. Saldo: {saldo_actual}, "
            f"cantidad solicitada: {retiro}"
        )
````

````python
def retirar(saldo, retiro):
    if saldo < retiro:
        raise SaldoInsuficienteError(saldo, retiro)

    return saldo - retiro

try:
    retirar(100 , 150)
except SaldoInsuficienteError as e:
    print(e)
````
